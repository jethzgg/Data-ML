```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")

np.random.seed(42)
def create_data():
    data = {
        "Age": np.random.randint(18, 70, size=20),
        "Salary": np.random.randint(30000, 120000, size=20),
        "Purchased": np.random.choice([0, 1], size=20),
        "Gender": np.random.choice(["Male", "Female"], size=20),
        "City": np.random.choice(["New York", "San Francisco", "Los Angeles"], size=20)
    }

    df = pd.DataFrame(data)
    return df

df = create_data()
df.head(10)
```

# When you have missing value in data

```python
# Introduce some missing values
df.loc[5, "Age"] = np.nan
df.loc[10, "Salary"] = np.nan
df.head(11)
```

### Simple Imputation (mean, median, mode)
Pros: Fast and straight forward.

Cons: May introduce bias if missing data is not random.

```python
df["Age"].fillna(df["Age"].mean(), inplace=True)
df["Salary"].fillna(df["Salary"].median(), inplace=True)
df.head(11)
```

### Listwise Deletion
Pros: Simple.

Cons: Loosing data.

```python
df_dropped = df.dropna()
df_dropped.head(11)
```

### Predictive Imputation - Search for MICE imputation as well
Pros: More accurate for complex patterns.

Cons: Computationally intensive, may add noise with high variance.

```python
from sklearn.impute import KNNImputer
imputer = KNNImputer(n_neighbors=3)
df[["Age", "Salary"]] = imputer.fit_transform(df[["Age", "Salary"]])
df.head(11)
```

### Add indicator for missingness
Pros: Allows model to learn from missingness.

Cons: Can invrease dimensionality and complexity.

```python
df["Age_missing"] = df["Age"].isnull().astype(int)
df.head(11)
```

# When you have Categorical Variables in data

### Label Encoding
Pros: Simple and space-efficient.

Cons: Implies ordinal relationship, which may mislead the model.

```python
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
df["Gender"] = label_encoder.fit_transform(df["Gender"])
df.head()
```

### One-hot Encoding
Pros: Great for non-ordinal categories, maintain all category information.

Cons: Increases dimensionality, particularly with many unique values.

```python
df = pd.get_dummies(df, columns=["City"], drop_first=True)
df.head()
```

### Ordinal Encoding
Pros: Effective for ordinal data, captures order.

Cons: Not suitable for non-ordinal data, as it implies ranking.

```python
df["City"] = df["City"].map({"New York": 1, "San Francisco": 2, "Los Angeles": 3})
df.head()
```

### Target Encoding
Pros: Useful for high-cardinality categorical variables.

Cons: Can cause data leakage if target encoding is not done properly.

```python
mean_target = df.groupby("City")["Purchased"].mean()
df["City_target_encoded"] = df["City"].map(mean_target)
df.head()
```

#### When to use
- Use label encoding for ordinal categorical variables.
- Use one-hot encoding for nominal categorical variables with low cardinality.
- Use ordinal encoding for ordinal categorical variables.
- Use target encoding for high-cardinality categorical variables, but ensure to avoid data leakage by applying it only on training data and using cross-validation.

# When you need to scale your data

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler
```

### Standardization (mean=0, variance=1)
Pros: Works well for normally distributed data, suitable for many models.

Cons: Sensitive to outliers.

```python
scaler = StandardScaler()
df[["Age", "Salary"]] = scaler.fit_transform(df[["Age", "Salary"]])
df.head()
```

### Normalization (range [0,1])
Pros: Keeps all data between 0 and 1, ideal for distance-based models.

Cons: Can distort data distribution, sensitive to outliers.

```python
normalizer = MinMaxScaler()
df[["Age", "Salary"]] = normalizer.fit_transform(df[["Age", "Salary"]])
df.head()
```

### Robust Scaling (median=0, IQR=1, robust to outliers)
Pros: More robust to outliers by using median and IQR.

Cons: Does not work as well for normal distributions.

```python
robust_scaler = RobustScaler()
df[["Age", "Salary"]] = robust_scaler.fit_transform(df[["Age", "Salary"]])
df.head()
```

#### When to use
- Use standardization for normally distributed data or when using models that assume normality and outliers are cleaned.
- Use normalization for distance-based models, when features have different scales or you need to transform it to 0 and 1 scale.
- Use robust scaling when data contains outliers and you want to minimize their impact on the scaling process.


# When you have outliers in data

### Outlier Detection and Treatment using IQR
Pros: Simple and effective for mild outliers.

Cons: May overly reduce variation if there are many extreme outliers.

```python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
# 1.5 is a common multiplier for IQR to detect outliers
df["Salary"] = np.where(df["Salary"] > upper_bound, upper_bound, 
                        np.where(df["Salary"] < lower_bound, lower_bound, 
                                 df["Salary"]))
df.head()
```

### Z-score method
Pros: Effective for normally distributed data.

Cons: Not suitable for non-normal distributions, may miss outliers in skewed distributions.

```python
from scipy import stats
df["Salary_zscore"] = stats.zscore(df["Salary"])
df["Salary"] = np.where(df["Salary_zscore"].abs() > 3, np.nan, df["Salary"])
# 3 is a common threshold
df.head()
```

### Median replacement for outliers
Pros: Keep distribution shape intact, useful when capping is not feasible.

Cons: May distort data if outliers represent real phenomena, not errors.

```python
df["Salary_zscore"] = stats.zscore(df["Salary"])
median_salary = df["Salary"].median()
df["Salary"] = np.where(df["Salary_zscore"].abs() > 3, median_salary, df["Salary"])
df.head()
```

#### When to use
- Outlier detection and treatment using IQR is suitable for mild outliers.
- Z-score method is effective for normally distributed data.
- Median replacement for outliers is useful when capping is not feasible and you want to maintain the distribution shape intact.

# Feature engineering

### Creating a new feature based on Salary (Discretization)
Pros: Simplified continuous features; useful for non-linear models.

Cons: Can lose data granularity, potentially reducing model accuracy.

```python
df["Income_Level"] = pd.cut(df["Salary"], bins=[0, 50000, 100000, 150000], labels=["Low", "Medium", "High"])
df.head(10)
```

# Polynomial Features
Pros: Captures complex relationships between features, can improve model performance.

Cons: Increases dimensionality, can lead to overfitting if not regularized properly.

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=2, include_bias=False)
poly_features = poly.fit_transform(df[["Age", "Salary"]])
df_poly = pd.DataFrame(poly_features, columns=poly.get_feature_names_out(["Age", "Salary"]))
df = pd.concat([df, df_poly], axis=1)
df.head()
```

#### When to use
- Use Discretization when you want to simplify continuous features and capture non-linear relationships.
- Use Polynomial Features when you want to capture complex relationships between features, but be cautious of overfitting and increased dimensionality.
