# Adding / Removing Columns

```python
coffee['price'] = 4.99
```

```python
import numpy as np

coffee['new price'] = np.where(coffee['Coffee Type'] == 'Espresso', 3.99, 5.99)
```

```python
coffee.drop(columns=['price'], inplace=True)
# inplace=True means that the original DataFrame will be modified and the 'price' column will be removed from it. If inplace=False (or not specified), a new DataFrame will be returned with the 'price' column removed, and the original DataFrame will remain unchanged.
```

```python
coffee = coffee[['Day', 'Coffee Type', 'Units Sold', 'new price']]
```

```python
coffee['revenue'] = coffee['Units Sold'] * coffee['new price']
```

```python
coffee.rename(columns={'new price': 'price'}, inplace=True)
```

# Aggregating Data

```python
bios[bios['born_country'] == 'USA']['born_region'].value_counts()
# value_counts() returns a Series containing counts of unique values. The resulting object will be in descending order so that the first element is the most frequently-occurring element.
```

```python
coffee.groupby('Coffee Type')['Units Sold'].sum()
```

```python
coffee.groupby(['Coffee Type', 'Day']).agg({'Units Sold': 'sum', 'price': 'mean'})
# agg means aggregate. It allows you to apply multiple aggregation functions to different columns of a DataFrame. In this case, we are calculating the sum of 'Units Sold' and the mean of 'price' for each 'Coffee Type'.
```

```python
pivot = coffee.pivot(columns='Coffee Type', index='Day', values='revenue')
```

```python
pivot.sum()
# axis = 0: row
# axis = 1: column
```

```python
bios['born_date'] = pd.to_datetime(bios['born_date'])
bios.groupby(bios['born_date'].dt.year)['name'].count().reset_index().sort_values('name', ascending=False)
```

```python
bios['born_date'] = pd.to_datetime(bios['born_date'])
bios['born_year'] = bios['born_date'].dt.year
bios['born_month'] = bios['born_date'].dt.month
bios.groupby([bios['born_year'], bios['born_month']])['name'].count().reset_index().sort_values('name', ascending=False)
```

# Advanced Functionality

```python
coffee['yesterday_revenue'] = coffee['revenue'].shift(2)
```

```python
coffee['pct_change'] = coffee['revenue'] / coffee['yesterday_revenue'] * 100
```

```python
coffee
```

```python
bios['height_rank'] = bios['height_cm'].rank()
```

```python
bios.sort_values('height_rank', ascending=True)[['name', 'height_rank']]
```

```python
coffee['cumulative_revenue'] = coffee['revenue'].cumsum()
```

```python
latte = coffee[coffee['Coffee Type'] == 'Latte'].copy()
latte['3 days'] = latte['Units Sold'].rolling(3).sum()
```
